```python
import sklearn
sklearn.set_config(transform_output="pandas")

from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV, ParameterGrid

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer

from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
```

```python
categorical_trans=Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder",OneHotEncoder(handle_unknown="ignore",sparse_output=False))
    ]
)
```

```python
categorical_trans
```

```python
numeric_trans=Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scalar",MinMaxScaler())
    ]
)
```

```python
numeric_trans
```

```python
cat_feat = ['standard_make', 'standard_model','vehicle_condition','body_type','fuel_type']
num_feat = ['year_of_registration', 'vehicle_age', 'mileage']
```

```python
preproc=ColumnTransformer(
    transformers=[
        ("cateo",categorical_trans,cat_feat),
        ("num",numeric_trans,num_feat)
    ]
)
```

```python
preproc
```

```python
# Define features (X) and target (y)
X = df.drop(columns=['log_price'])  # Exclude the target column
y = df['log_price']  # Define the target

# Split into train, validation, and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("\nData splits:")
print(f"Training set: {X_train.shape},Test set: {X_test.shape}")
```

KNN Regressor

```python
from sklearn.neighbors import KNeighborsRegressor
knn_reg=Pipeline([
    ("preprocessor",preproc),
    ("knnregressor",KNeighborsRegressor(n_neighbors=8))
])
knn_reg.fit(X_train,y_train)
```

```python
train_score=knn_reg.score(X_train, y_train)
```

```python
train_score
```

```python
y_knn = knn_reg.predict(X_test)
```

```python
y_knn
```

```python
param_grid = {
    'knnregressor__n_neighbors': [3, 5, 6],
    'knnregressor__weights': ['uniform', 'distance']
}
```

```python
from sklearn.model_selection import ParameterGrid

# Display a list of hyperparameter combinations
list(ParameterGrid(param_grid))
```

```python

# Initialize GridSearchCV with the updated param_grid
knn_grid = GridSearchCV(knn_reg, param_grid, cv=3,return_train_score=True)

# Fit the GridSearchCV on the scaled training data
knn_grid .fit(X_train, y_train)

# Access and display GridSearchCV results
gs_knn = pd.DataFrame(knn_grid.cv_results_)
print(gs_knn)
```
